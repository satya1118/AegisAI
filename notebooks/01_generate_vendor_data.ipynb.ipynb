```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

np.random.seed(42)

NUM_VENDORS = 500
TODAY = datetime.today()

categories = ["Cloud", "SaaS", "Logistics", "IT Services", "Security", "Finance"]

vendors = []

for i in range(NUM_VENDORS):
    vendor_id = f"V_{1000+i}"
    vendor_name = f"Vendor_{i+1}"
    category = np.random.choice(categories)

    contract_value = np.random.randint(50000, 5000000)

    contract_start = TODAY - timedelta(days=np.random.randint(30, 900))
    contract_end = contract_start + timedelta(days=np.random.randint(180, 1095))

    sla_breaches_90d = np.random.poisson(1.5)
    incidents_90d = np.random.poisson(1.0)

    compliance_flag = np.random.choice([0, 1], p=[0.8, 0.2])
    uptime_pct = round(np.random.normal(99, 1.5), 2)
    uptime_pct = max(90, min(uptime_pct, 100))

    financial_risk_score = round(np.random.beta(2, 5), 2)

    risk_probability = (
        sla_breaches_90d * 0.25 +
        incidents_90d * 0.20 +
        compliance_flag * 0.30 +
        (100 - uptime_pct) * 0.15 +
        financial_risk_score * 0.10
    )

    if risk_probability < 0.8:
        risk_tier = "Low"
    elif risk_probability < 1.6:
        risk_tier = "Medium"
    else:
        risk_tier = "High"

    vendor_exit_risk = 1 if risk_tier == "High" else 0

    vendors.append([
        vendor_id, vendor_name, category, contract_value,
        contract_start.date(), contract_end.date(),
        sla_breaches_90d, incidents_90d,
        compliance_flag, uptime_pct,
        financial_risk_sco
        round(risk_probability, 2),
        risk_tier,
        vendor_exit_risk
    ])

columns = [
    "vendor_id", "vendor_name", "category", "contract_value",
    "contract_start", "contract_end",
    "sla_breaches_90d", "incidents_90d",
    "compliance_flag", "uptime_pct",
    "financial_risk_score",
    "risk_probability", "risk_tier", "vendor_exit_risk"
]

df = pd.DataFrame(vendors, columns=columns)

df.to_csv("vendors.csv", index=False)

print("Dataset created successfully")
```
